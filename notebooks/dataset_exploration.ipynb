# Dataset Exploration — Synthetic Complaint Sample
This notebook explores the small illustrative dataset included with the feasibility proof. It is synthetic and is not evidence about real municipal complaints.

```python
import pandas as pd
from pathlib import Path
df = pd.read_csv('../data/synthetic_complaints.csv')
df.head()
```

```python
print('Rows:', len(df))
print('\nMissing values by column:')
print(df.isna().sum())
print('\nChannel counts:')
print(df['channel'].value_counts())
print('\nDepartment counts:')
print(df['department'].value_counts())
```

```python
import matplotlib.pyplot as plt
df['department'].value_counts().plot(kind='bar', title='Synthetic examples by department')
plt.xlabel('Department')
plt.ylabel('Number of examples')
plt.tight_layout()
plt.show()
```

## Interpretation
The sample is deliberately small and balanced enough to demonstrate the pipeline. It is useful for checking data loading, fields and basic plots, but too small to support claims about real-world accuracy or generalisation. Duplicate labels are illustrative and should be reviewed before model evaluation.
